```python
dados = pd.concat([marca_a_padrao, marca_b_padrao], ignore_index=True)


por_cliente = dados.groupby(["marca", "customer_id"]).agg(
    ultima_compra=("data_pedido", "max"),
    primeira_compra=("data_pedido", "min"),
    total_pedidos=("order_id", "count"),
    total_gasto=("valor_total", "sum"),
).reset_index()

por_cliente.head()
```

```python
por_cliente.groupby("marca")["dias_desde_ultima_compra"].describe()
```

```python
LIMITE_PERCENTIL = 0.75

por_cliente["limite_churn_dias"] = por_cliente.groupby("marca")["dias_desde_ultima_compra"].transform(
    lambda x: x.quantile(LIMITE_PERCENTIL)
)
por_cliente["churn"] = (por_cliente["dias_desde_ultima_compra"] > por_cliente["limite_churn_dias"]).astype(int)

por_cliente.groupby("marca")["churn"].value_counts(normalize=True)
```
